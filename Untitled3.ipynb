```python
df = pd.read_csv("pool.csv", sep=";", decimal=",", header=None)

df.columns = ["PoolID", "PoolName", "Length", "Width", "Depth"]

df.head()
```

```python
length_counts = df["Length"].value_counts().reset_index()
length_counts.columns = ["Length", "Count"]

plt.figure(figsize=(8,5))
sns.barplot(data=length_counts, x="Length", y="Count")
plt.title("Количество бассейнов по длине")
plt.xlabel("Length")
plt.ylabel("Количество")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df["Width"], bins=20)
plt.title("Распределение ширины бассейнов")
plt.xlabel("Width")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x="Width", y="Depth")
plt.title("Связь ширины и глубины")
plt.xlabel("Width")
plt.ylabel("Depth")
plt.tight_layout()
plt.show()
```
