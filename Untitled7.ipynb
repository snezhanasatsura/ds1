```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import TimeSeriesSplit
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

from sklearn.linear_model import LinearRegression, Ridge
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor, ExtraTreesRegressor, IsolationForest
from sklearn.preprocessing import StandardScaler

import warnings
warnings.filterwarnings("ignore")

sns.set(style="whitegrid")
```

```python
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/00616/Tetuan%20City%20power%20consumption.csv"
df = pd.read_csv(url)
df.head()
```

```python
print("Размер датасета:", df.shape)
print()
df.info()
print()
print(df.isnull().sum())
```

В работе используется датасет Tetuan City power consumption, содержащий данные о погодных условиях и потреблении электроэнергии в трёх зонах города.

Описание признаков:

DateTime — дата и время наблюдения

Temperature — температура воздуха

Humidity — влажность

Wind Speed — скорость ветра

general diffuse flows — общий рассеянный поток

diffuse flows — диффузный поток

Zone 1 Power Consumption — потребление электроэнергии в зоне 1

Zone 2 Power Consumption — потребление электроэнергии в зоне 2

Zone 3 Power Consumption — потребление электроэнергии в зоне 3

В данной работе в качестве целевой переменной используется Zone 1 Power Consumption.

```python
df['DateTime'] = pd.to_datetime(df['DateTime'])
df = df.sort_values('DateTime')
df.head()
```

```python
df.columns = [
    'datetime',
    'temperature',
    'humidity',
    'wind_speed',
    'general_diffuse_flows',
    'diffuse_flows',
    'zone_1',
    'zone_2',
    'zone_3'
]

df.head()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['zone_1'], bins=30, kde=True)
plt.title("Распределение потребления электроэнергии (Zone 1)")
plt.xlabel("Zone 1 Power Consumption")
plt.show()
```

```python
plt.figure(figsize=(8,4))
sns.boxplot(x=df['zone_1'])
plt.title("Boxplot потребления электроэнергии (Zone 1)")
plt.show()
```

```python
df['hour'] = df['datetime'].dt.hour
df['day'] = df['datetime'].dt.day
df['month'] = df['datetime'].dt.month
df['dayofweek'] = df['datetime'].dt.dayofweek
df['is_weekend'] = df['dayofweek'].isin([5, 6]).astype(int)
```

```python
plt.figure(figsize=(10,5))
sns.lineplot(x='hour', y='zone_1', data=df)
plt.title("Среднее потребление по часам")
plt.xlabel("Час")
plt.ylabel("Потребление")
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.barplot(x='dayofweek', y='zone_1', data=df)
plt.title("Среднее потребление по дням недели")
plt.xlabel("День недели (0=понедельник)")
plt.ylabel("Потребление")
plt.show()
```

```python
df['lag_1'] = df['zone_1'].shift(1)
df['lag_2'] = df['zone_1'].shift(2)
df['lag_3'] = df['zone_1'].shift(3)
df['lag_24'] = df['zone_1'].shift(24)

df['rolling_mean_3'] = df['zone_1'].shift(1).rolling(window=3).mean()
df['rolling_mean_24'] = df['zone_1'].shift(1).rolling(window=24).mean()
df['rolling_std_24'] = df['zone_1'].shift(1).rolling(window=24).std()
```

```python
df = df.dropna().reset_index(drop=True)
print(df.shape)
```

```python
features = [
    'temperature', 'humidity', 'wind_speed',
    'general_diffuse_flows', 'diffuse_flows',
    'hour', 'day', 'month', 'dayofweek', 'is_weekend',
    'lag_1', 'lag_2', 'lag_3', 'lag_24',
    'rolling_mean_3', 'rolling_mean_24', 'rolling_std_24'
]

target = 'zone_1'

X = df[features]
y = df[target]
```

```python
split_index = int(len(df) * 0.8)

X_train = X.iloc[:split_index]
X_test = X.iloc[split_index:]

y_train = y.iloc[:split_index]
y_test = y.iloc[split_index:]

dates_train = df['datetime'].iloc[:split_index]
dates_test = df['datetime'].iloc[split_index:]

print("Train:", X_train.shape)
print("Test:", X_test.shape)
```

```python
scaler = StandardScaler()

X_train_scaled = X_train.copy()
X_test_scaled = X_test.copy()

X_train_scaled[features] = scaler.fit_transform(X_train[features])
X_test_scaled[features] = scaler.transform(X_test[features])
```

```python
models = {
    'Linear Regression': LinearRegression(),
    'Ridge Regression': Ridge(alpha=1.0),
    'Random Forest': RandomForestRegressor(n_estimators=150, random_state=42, n_jobs=-1),
    'Gradient Boosting': GradientBoostingRegressor(n_estimators=150, random_state=42),
    'Extra Trees': ExtraTreesRegressor(n_estimators=150, random_state=42, n_jobs=-1)
}
```

```python
def evaluate_model(name, model, X_train, X_test, y_train, y_test):
    model.fit(X_train, y_train)
    preds = model.predict(X_test)

    mae = mean_absolute_error(y_test, preds)
    rmse = np.sqrt(mean_squared_error(y_test, preds))
    r2 = r2_score(y_test, preds)
    mape = np.mean(np.abs((y_test - preds) / y_test)) * 100

    return {
        'Model': name,
        'MAE': mae,
        'RMSE': rmse,
        'R2': r2,
        'MAPE': mape,
        'Predictions': preds
    }
```

```python
results = []

# Linear Regression
results.append(
    evaluate_model('Linear Regression', models['Linear Regression'],
                   X_train_scaled, X_test_scaled, y_train, y_test)
)

# Ridge Regression
results.append(
    evaluate_model('Ridge Regression', models['Ridge Regression'],
                   X_train_scaled, X_test_scaled, y_train, y_test)
)

# Random Forest
results.append(
    evaluate_model('Random Forest', models['Random Forest'],
                   X_train, X_test, y_train, y_test)
)

# Gradient Boosting
results.append(
    evaluate_model('Gradient Boosting', models['Gradient Boosting'],
                   X_train, X_test, y_train, y_test)
)

# Extra Trees
results.append(
    evaluate_model('Extra Trees', models['Extra Trees'],
                   X_train, X_test, y_train, y_test)
)
```

```python
results_df = pd.DataFrame(results)[['Model', 'MAE', 'RMSE', 'R2', 'MAPE']]
results_df = results_df.sort_values(by='RMSE')
results_df
```

```python
plt.figure(figsize=(10,5))
sns.barplot(data=results_df, x='Model', y='RMSE')
plt.xticks(rotation=45)
plt.title("Сравнение моделей по RMSE")
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.barplot(data=results_df, x='Model', y='R2')
plt.xticks(rotation=45)
plt.title("Сравнение моделей по R2")
plt.show()
```

```python
best_model_name = results_df.iloc[0]['Model']
best_model_name
```

```python
best_result = next(r for r in results if r['Model'] == best_model_name)
best_preds = best_result['Predictions']
```

```python
plt.figure(figsize=(16,5))
plt.plot(dates_test.values[:300], y_test.values[:300], label='Actual')
plt.plot(dates_test.values[:300], best_preds[:300], label='Predicted')
plt.title(f"Фактические и прогнозные значения: {best_model_name}")
plt.xlabel("Дата")
plt.ylabel("Потребление")
plt.legend()
plt.show()
```

```python
errors = y_test.values - best_preds

plt.figure(figsize=(16,5))
plt.plot(dates_test.values[:300], errors[:300])
plt.axhline(0, color='red', linestyle='--')
plt.title("Ошибки прогнозирования")
plt.xlabel("Дата")
plt.ylabel("Ошибка")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(errors, bins=30, kde=True)
plt.title("Распределение ошибок прогнозирования")
plt.xlabel("Ошибка")
plt.show()
```

```python
plt.figure(figsize=(7,7))
plt.scatter(y_test, best_preds, alpha=0.5)
plt.xlabel("Реальные значения")
plt.ylabel("Предсказанные значения")
plt.title("Реальные vs предсказанные значения")
plt.show()
```

```python
if best_model_name == 'Random Forest':
    best_model = models['Random Forest']
    best_model.fit(X_train, y_train)
elif best_model_name == 'Gradient Boosting':
    best_model = models['Gradient Boosting']
    best_model.fit(X_train, y_train)
elif best_model_name == 'Extra Trees':
    best_model = models['Extra Trees']
    best_model.fit(X_train, y_train)
elif best_model_name == 'Linear Regression':
    best_model = models['Linear Regression']
    best_model.fit(X_train_scaled, y_train)
elif best_model_name == 'Ridge Regression':
    best_model = models['Ridge Regression']
    best_model.fit(X_train_scaled, y_train)
```

```python
if hasattr(best_model, 'feature_importances_'):
    importances = pd.DataFrame({
        'Feature': features,
        'Importance': best_model.feature_importances_
    }).sort_values(by='Importance', ascending=False)

    plt.figure(figsize=(10,6))
    sns.barplot(data=importances.head(10), x='Importance', y='Feature')
    plt.title("Топ-10 наиболее важных признаков")
    plt.show()

    importances.head(10)
```

```python
anomaly_df = pd.DataFrame({
    'datetime': dates_test.values,
    'actual': y_test.values,
    'predicted': best_preds,
    'error': errors,
    'abs_error': np.abs(errors)
})
```

```python
iso = IsolationForest(contamination=0.03, random_state=42)

anomaly_df['anomaly'] = iso.fit_predict(anomaly_df[['actual', 'predicted', 'abs_error']])
anomaly_df['anomaly_label'] = anomaly_df['anomaly'].map({1: 'normal', -1: 'anomaly'})

anomaly_df.head()
```

```python
plt.figure(figsize=(16,5))
plt.plot(anomaly_df['datetime'], anomaly_df['actual'], label='Actual', alpha=0.7)
plt.scatter(
    anomaly_df[anomaly_df['anomaly_label'] == 'anomaly']['datetime'],
    anomaly_df[anomaly_df['anomaly_label'] == 'anomaly']['actual'],
    color='red',
    label='Anomalies',
    s=20
)
plt.title("Аномалии в энергопотреблении")
plt.xlabel("Дата")
plt.ylabel("Потребление")
plt.legend()
plt.show()
```
